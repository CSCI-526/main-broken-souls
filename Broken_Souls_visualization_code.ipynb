```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv('analytics.csv', parse_dates=['Timestamp'], dayfirst=False)


df['Survival Time'] = pd.to_numeric(df['Survival Time'], errors='coerce')
df = df.dropna(subset=['Survival Time'])


plt.figure(figsize=(10,6))
n_bins = 30
plt.hist(df['Survival Time'], bins=n_bins, edgecolor='black')
plt.xlabel('Survival time (seconds)')
plt.ylabel('Number of plays')
plt.title('Distribution of Player Survival Time')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('survival_time_histogram.png', dpi=150)
plt.show()
```
